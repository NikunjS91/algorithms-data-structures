# Binary Search



# binary search can be only applied over a sortted arrau, either ascending or descending array


```python
def binarySearch(arr,target):
    size=len(arr)
    start=0
    end = size-1
    while(start<=end):
        mid=int(start+(end-start)/2)
        if(target > arr[mid]):
            start=mid+1
        elif(target<arr[mid]):
            end= mid-1
        else:
            return mid
    return "not in the list"

array=[1,2,3,4,5]
print("pick element for search : ",array)
target=int(input(" : "))
result = binarySearch(array,target)
print(f"{array[result]} is at {result} index")
```
