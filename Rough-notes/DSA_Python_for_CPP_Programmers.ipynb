# Python for C++ Programmers: DSA Essentials
This notebook summarizes the mapping between C++ STL containers/logic and Python built-ins, specifically for Data Structures and Algorithms (DSA).

## 1. Pointers and References
In Python, variables are labels (references) to objects in memory. While we don't have pointer arithmetic, `id()` gives us the address.

```python
x = [1, 2, 3]
y = x  # y points to the same address as x
print(f"Address of x: {hex(id(x))}")
print(f"Address of y: {hex(id(y))}")
print(f"x is y: {x is y}") # Checks if they share the same memory address
```

## 2. Dynamic Arrays (`std::vector` -> `list`)
Python lists are dynamic arrays. 
- `push_back` -> `.append()`
- `pop_back` -> `.pop()`
- `back()` -> `[-1]`
- `vector<int> v(size, val)` -> `[val] * size`

```python
vec = [1, 2, 3]
vec.append(4)  # push_back
last_element = vec[-1] # back()
vec.pop() # pop_back
fill_vec = [10] * 5 # vector<int> v(5, 10)
print(f"Vector: {vec}, Filled: {fill_vec}")
```

## 3. Double-Ended Queue (`std::deque` -> `collections.deque`)
Essential for $O(1)$ front operations in DSA.

```python
from collections import deque
dq = deque([1, 2, 3])
dq.appendleft(0) # push_front
dq.popleft()     # pop_front
print(f"Deque: {dq}")
```

## 4. Pairs and Tuples (`std::pair` -> `tuple`)
Python tuples are immutable pairs/groups.

```python
pair = (10, "Apple")
first = pair[0]  # p.first
second = pair[1] # p.second

# Unpacking (Very useful in DSA loops)
a, b = pair
print(f"Unpacked: {a}, {b}")
```

## 5. DSA Example: Peak Index in Mountain Array
Implementing binary search logic using Python's floor division `//`.

```python
def peakIndex(arr):
    start, end = 1, len(arr) - 2
    while start <= end:
        mid = start + (end - start) // 2
        if arr[mid-1] < arr[mid] > arr[mid+1]:
            return mid
        if arr[mid-1] < arr[mid]:
            start = mid + 1
        else:
            end = mid - 1
    return -1

arr = [0, 3, 8, 9, 5, 2]
print(f"Peak index is: {peakIndex(arr)}")
```
