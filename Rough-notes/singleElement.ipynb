```python
def singleElement(arr):
    size = len(arr)
    start=0
    end=size-1
    if(size==1):return arr[0]
    while(start<=end):
        mid = start+(end-start)//2
        if(mid==0 and arr[0]!=arr[1]):
            return arr[mid]
        if(mid==size-1 and arr[size-1]!=arr[size-2]):
            return mid
        if(arr[mid]!=arr[mid+1] and arr[mid]!= arr[mid-1] ):
            return arr[mid]
        if(mid%2==0):
            if(arr[mid]== arr[mid-1]):
                end=mid-1
        else:
            if(arr[mid]== arr[mid-1]):
                start=mid+1

            
    

arr=[1,1,2,3,3,4,4,8,8]
result=singleElement(arr)
print(result)
```
