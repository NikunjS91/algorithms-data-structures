# Product of Array Except Self

```python
def prodArray(arr):
    size =len(arr)
    new_array=[]
    for i in range(size):
        prod=1
        for j in range(size):
            if(j!=i):
                prod*=arr[j]
        new_array.append(prod)
    return new_array

arr=[1,2,3,4]
result = prodArray(arr)
print(result)
```
