```python
def isAlphaNum(ch):
    if((ch>="0" and ch<="9") or (ch.lower() >="a" and ch.lower()<="z")):
        return True
    return False



def isPalindrome(s):
    start=0
    end=len(s)-1
    while(start<end):
        if not isPalindrome(s[start]):
           start+=1
           continue
        if not isAlphaNum(s[end]):
           end-=1
           continue
        if(s[start].lower() != s[end].lower()):
            return False
        start+=1
        end-=1
    return True
           
           



s="Ac3?e3c&a"
result= isPalindrome(s)
if(result):
    print(f"{s} is palindrome")
else:
    print(f"{s} is not palindrome")
```
