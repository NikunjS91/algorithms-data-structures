```python
def painterFunction(arr,size,painters):
    start=0
    end=(sum(arr))
    ans=-1
    while(start<=end):
        mid=start+(end-start)//2
        if(isValid(arr,size, mid,painters)):
            end=mid-1
            ans=mid
        else:
            start=mid+1
    return ans

def isValid(arr,size,mid,painters):
    person=1
    board=0
    for i in range(size):
        if(arr[i]+board<=mid):
            board+=arr[i]
        else:
            person+=1
            board=arr[i]
    if(person>painters):
        return False
    if(person<=painters):
        return True




arr = [40,30,10,20]
size=len(arr)
painters=2


result= painterFunction(arr,size,painters)
print(result)
```
