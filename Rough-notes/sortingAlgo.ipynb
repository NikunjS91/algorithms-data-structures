```python
def bubbleSort(arr,size): #n^2
    for i in range(size):#n
        for j in range(size-i-1):#n*n
            if(arr[j]>arr[j+1]):
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr    


arr=[4,1,5,2,3]
size= len(arr)
result=bubbleSort(arr,size)
print(result)
```

```python
def selectionSort(arr,size):
    for i in range(size-1):
        smallestElement=i
        for j in range(i+1,size,+1):
            if(arr[j]<arr[smallestElement]):
                smallestElement=j
        arr[smallestElement],arr[i]=arr[i],arr[smallestElement]
    return arr
arr=[4,1,5,2,3]
size= len(arr)
result=selectionSort(arr,size)
print(result)
```

```python
def insersionSort(arr,size):
    for i in range(1,size):
        curr=arr[i]
        previous=i-1
        while(previous>=0 and arr[previous]>curr):
            arr[previous+1]=arr[previous]
            previous-=1
        arr[previous+1]=curr
    return arr

arr=[4,1,5,2,3]
size= len(arr)
result=insersionSort(arr,size)
print(result)
```
