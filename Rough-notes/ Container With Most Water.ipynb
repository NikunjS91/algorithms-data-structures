```python
#brute force
def water(arr):
    arr_size=len(arr)
    max_water=0
    for i in range(arr_size):
        for j in range(i+1,arr_size):
            width=j-i
            height=min(arr[j],arr[i])
            area = height*width
            max_water=max(max_water,area)
    return max_water

height = [1,8,6,2,5,4,8,3,7]
result = water(height)
print("max water is : ",result)
```

```python
# optimal two pointer
def water1(arr):
    arr_size = len(arr)
    max_water= 0
    left = 0
    right = arr_size-1
    while(left<right):
        width =right-left
        height=min(arr[left],arr[right])
        area=width*height
        max_water=max(max_water,area)
        if(arr[left]<arr[right]):
            left+=1
        else:
            right-=1
    return max_water

height = [1,8,6,2,5,4,8,3,7]
result = water1(height)
print("max water in optimal solution is  : ",result)
```
