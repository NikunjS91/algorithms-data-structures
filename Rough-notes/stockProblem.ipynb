# Compute x^n

```python
def myPow(x,n):
    if(n==0): return 1.0
    if(x==0): return 0.0
    if(x==1): return 1.0
    if(x==-1 and n%2==0): return 1.0
    if(x==-1 and n%2!=0): return -1.0
    
    binaryForm=n
    if(n<0):
        x=1/x
        binaryForm=-binaryForm
    ans=1
    while(binaryForm>0):
        if(binaryForm%2==1):
            ans*=x 
        x*=x
        binaryForm//=2
    return ans

result=myPow(2.00000,10)
print(result)
```

# Stock Buy and Sell


```python
def profit(arr):
    max_profit=0
    best_buy=arr[0]
    size =len(arr)
    for i in range(1,size):
        if(arr[i]>best_buy):
            max_profit=max(max_profit,arr[i]-best_buy)
        best_buy=min(best_buy,arr[i])
    return max_profit

arr=[7,1,5,3,6,4]
result=profit(arr)
print("max profit is :",result)
```
