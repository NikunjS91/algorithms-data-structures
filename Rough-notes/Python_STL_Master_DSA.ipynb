# Master Python for C++ Programmers: Final STL Roadmap
This notebook covers the final set of STL-to-Python mappings: Stacks, Queues, Priority Queues, Maps, Sets, and Advanced Algorithms.

## 1. Stack (LIFO)
**STL Equivalent:** `std::stack`  
**Python tool:** `list` (using `append` and `pop`)
- `push`/`emplace` -> `.append()`
- `top` -> `[-1]`
- `pop` -> `.pop()`

```python
stack = []
stack.append(10) # push
stack.append(20)
print(f"Top: {stack[-1]}") # top
stack.pop() # pop
print(f"Empty? {not stack}") # empty()
```

## 2. Queue (FIFO)
**STL Equivalent:** `std::queue`  
**Python tool:** `collections.deque`  
*Note: Do not use `list.pop(0)` as it is $O(n)$.*

```python
from collections import deque
q = deque()
q.append(1) # push
q.append(2)
print(f"Front: {q[0]}") # front
q.popleft() # pop (removes from front in O(1))
print(f"Size: {len(q)}") # size
```

## 3. Priority Queue (Heap)
**STL Equivalent:** `std::priority_queue`  
**Python tool:** `heapq` module  
**CRITICAL:** C++ is a Max-Heap by default; Python is a **Min-Heap**.

```python
import heapq
pq = []
heapq.heappush(pq, 10)
heapq.heappush(pq, 5)
heapq.heappush(pq, 20)
print(f"Min-Heap Top: {pq[0]}") # 5

# To simulate Max-Heap, multiply by -1
max_pq = []
heapq.heappush(max_pq, -10)
heapq.heappush(max_pq, -20)
print(f"Max-Heap Top: {-max_pq[0]}") # 20
```

## 4. Map and Set
**STL Equivalents:** `unordered_map` -> `dict`, `unordered_set` -> `set`  
**Multimap** -> `defaultdict(list)`

```python
from collections import defaultdict

# Set (Unique elements)
s = {1, 2, 3}
s.add(4)
print(f"4 in set? {4 in s}")

# Multimap (Multiple values per key)
mm = defaultdict(list)
mm['key'].append('val1')
mm['key'].append('val2')
print(f"Multimap values: {mm['key']}")
```

## 5. Binary Search (Lower/Upper Bound)
**STL Equivalent:** `std::lower_bound` & `std::upper_bound`  
**Python tool:** `bisect` module

```python
import bisect
arr = [1, 2, 4, 4, 4, 6]
lb = bisect.bisect_left(arr, 4) # Index of first 4
ub = bisect.bisect_right(arr, 4) # Index after last 4
print(f"Lower Bound: {lb}, Upper Bound: {ub}")
```

## 6. Sorting and Custom Comparators
**STL Equivalent:** `std::sort(v.begin(), v.end(), comp)`

```python
# Sorting by a secondary attribute (e.g., sort pairs by second element)
pairs = [(1, 10), (2, 5), (3, 8)]
pairs.sort(key=lambda x: x[1])
print(f"Sorted by second value: {pairs}")

# Sorting in reverse
arr = [1, 5, 2]
arr.sort(reverse=True)
print(f"Descending: {arr}")
```
