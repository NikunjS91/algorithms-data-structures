```python
def reverseString(str):
    size=len(str)
    main_str=str[::-1]
    ans=""
    i=0
    while i<size:

        word = ""
        while(i<size and main_str[i]!=" "):
            word+=main_str[i]
            i+=1
        word=word[::-1]
        if(len(word)>0):
            ans+=" "+word
        i+=1
    return ans[1:]



str="my pen"
result=reverseString(str)
print(result)
```
