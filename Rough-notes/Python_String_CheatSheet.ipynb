# Strings and Character Arrays: C++ to Python
This notebook explains how C++ string concepts (including memory handling and I/O) translate to Python's high-level string objects.

---

## 1. The Null Value (`\0`)
In C++, `\0` marks the end of a string in a character array. 
In Python, strings are **objects** that store their own length. You don't need a null terminator.

| C++ | Python |
| :--- | :--- |
| `char arr[] = "Hi\0";` | `s = "Hi"` |
| `strlen(arr)` is 2 | `len(s)` is 2 |

## 2. Input/Output (`cin`, `getline`)
In C++, `cin` stops at spaces. `getline` reads the whole line.
In Python, `input()` always reads the whole line (like `getline`).

```python
# C++: cin >> str;
# Python equivalent (first word only):
# s = input().split()[0]

# C++: getline(cin, str);
s = "Hello World" # Simulating input()

# C++: getline(cin, str, ','); (Delimiter)
parts = "apple,banana,cherry".split(',')
print(f"First part: {parts[0]}")
```

## 3. Dynamic Runtime Resize & Concatenation
C++ strings resize their internal buffer. Python strings are **immutable** (cannot be changed). 
Adding to a string creates a **new** string in memory.

```python
s1 = "C++"
s2 = "Python"

# Concatenation
s3 = s1 + " vs " + s2
print(s3)

# DSA Tip: To efficiently 'resize' or build a string:
chars = []
for i in range(5):
    chars.append(str(i)) # O(1) append
final_str = "".join(chars) # O(N) join at once
print(f"Built string: {final_str}")
```

## 4. Comparisons and Length
Python uses `len()` for length and standard operators for lexicographical (alphabetical) comparison.

```python
s1 = "apple"
s2 = "banana"

print(f"Length: {len(s1)}") # length()
print(f"s1 < s2: {s1 < s2}")  # True (Lexicographical)
print(f"s1 == 'apple': {s1 == 'apple'}")
```

## 5. Loops and Reversing
Python provides several ways to iterate and a powerful shortcut for reversing.

```python
s = "Gemini"

# Loop by character
for char in s:
    print(char, end=" ")
print() 

# Loop by index
for i in range(len(s)):
    print(f"Index {i}: {s[i]}")

# Reverse a string (Slicing)
# C++: reverse(s.begin(), s.end())
reversed_s = s[::-1]
print(f"Reversed: {reversed_s}")
```
