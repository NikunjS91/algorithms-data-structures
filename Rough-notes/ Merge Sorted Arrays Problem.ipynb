# leetcode 88

```python
def mergeSortedArray(arr1,arr2,m,n):
    idx=m+n-1
    i=m-1
    j=n-1
    while(i>=0 and j>=0):
        if(arr2[j]>=arr1[i]):
            arr1[idx]=arr2[j]
            idx-=1
            j-=1
        else:
            arr1[idx]=arr1[i]
            idx-=1
            i-=1
        while(j>=0):
            arr1[idx]=arr2[j]
            idx-=1
            j-=1
    return arr1


arr1=[1,2,3,0,0,0]
arr2=[3,3,5]
m=3
n=3
result =mergeSortedArray(arr1,arr2,m,n)
print(result)
```

# leetcode 31

```python
def findPivit(arr,size):
    pivit =-1
    for i in range(size-2,0,-1):
        if(arr[i]<arr[i+1]):
            pivit=i
            break
    if(pivit==-1):
        arr.reverse()
        return arr

    for j in range(size-1,pivit,-1):
        if(arr[i]>arr[pivit]):
            arr[pivit],arr[i]= arr[i],arr[pivit]
            break
    start =pivit+1
    end=size-1
    while(start<=end):
        arr[j],arr[i]=arr[i],arr[j]
        start+=1
        end-=1
    return arr

arr=[1,2,5,4,3]
size =len(arr)
result = findPivit(arr,size)
print(result)

```
