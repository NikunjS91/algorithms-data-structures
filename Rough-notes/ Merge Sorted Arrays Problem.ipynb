```python
def mergeSortedArray(arr1,arr2,m,n):
    idx=m+n-1
    i=m-1
    j=n-1
    while(i>=0 and j>=0):
        if(arr2[j]>=arr1[i]):
            arr1[idx]=arr2[j]
            idx-=1
            j-=1
        else:
            arr1[idx]=arr1[i]
            idx-=1
            i-=1
        while(j>=0):
            arr1[idx]=arr2[j]
            idx-=1
            j-=1
    return arr1


arr1=[1,2,3,0,0,0]
arr2=[3,3,5]
m=3
n=3
result =mergeSortedArray(arr1,arr2,m,n)
print(result)
```
