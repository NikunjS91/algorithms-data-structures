## 1. Prime Number Check ($2 \rightarrow N$)
**Logic:** Instead of checking up to $N$, optimize your loop up to $\sqrt{N}$. If a number $N$ has a factor greater than $\sqrt{N}$, it must also have a matching factor smaller than $\sqrt{N}$.

In C++, you would write `for(int i = 2; i * i <= n; i++)`. In Python, you can write `while i * i <= n:` or use `range(2, int(n**0.5) + 1)`.

```python
def is_prime(n):
    if n <= 1:
        return False
    
    # Loop from 2 up to sqrt(n)
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False # Found a factor, not prime
            
    return True

# Print all primes from 2 to 30
primes_2_to_30 = [x for x in range(2, 31) if is_prime(x)]
print(f"Primes from 2 to 30: {primes_2_to_30}")
```

## 2. Sieve of Eratosthenes ($2 \rightarrow N$)
**Logic:** If you need to check primes for a large range up to $N$, checking each individually takes $O(N \sqrt{N})$. The Sieve optimizes this to $O(N \log(\log N))$ by creating a boolean array and marking out multiples of every prime found.

**C++ Mapping:** `vector<bool> prime(n + 1, true)` $\rightarrow$ `[True] * (n + 1)`

```python
def sieve_of_eratosthenes(n):
    # Create a boolean array initialized to True
    # Index represents the number
    prime_sieve = [True] * (n + 1)
    prime_sieve[0] = prime_sieve[1] = False # 0 and 1 are not prime
    
    for p in range(2, int(n**0.5) + 1):
        # If prime_sieve[p] is not changed, then it is a prime
        if prime_sieve[p]:
            # Update all multiples of p starting from p*p
            for i in range(p * p, n + 1, p):
                prime_sieve[i] = False
                
    # Gather all numbers that remained True
    primes = [i for i in range(2, n + 1) if prime_sieve[i]]
    return primes

print(f"Sieve primes up to 50: {sieve_of_eratosthenes(50)}")
```

## 3. Digit Extraction, Summation, & Counting
**Logic:** 
- `n % 10` extracts the last digit.
- `n // 10` slices off the last digit (Floor division / integer division equivalent to C++ `n /= 10`).

*CRITICAL:* In Python, `/` always yields a float (e.g., `5 / 2 = 2.5`). For digit extraction, you **must** use `//` to avoid running into floating-point precision issues.

```python
def process_digits(n):
    # Handle negative numbers for safety
    num = abs(n)
    
    digit_sum = 0
    digit_count = 0
    
    # Loop until the number becomes 0
    while num > 0:
        last_digit = num % 10   # Extract last digit
        digit_sum += last_digit # Accumulate sum
        digit_count += 1        # Increment count
        
        num = num // 10         # C++ equivalent to num /= 10
        
    return digit_count, digit_sum

test_num = 7824
count, total_sum = process_digits(test_num)
print(f"Number: {test_num}")
print(f"Digit Count: {count}")
print(f"Sum of Digits: {total_sum}")
```

## 4. Armstrong Number
**Logic:** An Armstrong number (like 153 or 371) is a number that equals the sum of its own digits each raised to the power of the total number of digits.

**Python Power Operator:** `base ** exp` (replaces `pow(base, exp)` from C++ `<cmath>`).

```python
def is_armstrong(n):
    # Step 1: Count total digits
    # Python Shortcut: count = len(str(n))
    # Let's do it mathematically to match standard math logic:
    temp = n
    num_digits = 0
    while temp > 0:
        num_digits += 1
        temp //= 10
        
    # Step 2: Sum each digit raised to num_digits
    temp = n
    armstrong_sum = 0
    while temp > 0:
        digit = temp % 10
        armstrong_sum += (digit ** num_digits) # Using Python's ** operator
        temp //= 10
        
    # Step 3: Compare against original number
    return armstrong_sum == n

print(f"Is 153 an Armstrong number? {is_armstrong(153)}") # 1^3 + 5^3 + 3^3 = 153
print(f"Is 123 an Armstrong number? {is_armstrong(123)}")
```

## 5. Greatest Common Divisor (GCD / HCF)
**Logic:** The Euclidean Algorithm states that the GCD of two numbers also divides their difference. Efficiently computed via recursively or iteratively running `gcd(b, a % b)` until the remainder becomes 0.

Time Complexity: $O(\log(\min(a, b)))$

```python
def get_gcd(a, b):
    # Iterative Euclidean Algorithm
    while b > 0:
        a, b = b, a % b # Python simultaneous assignment swaps and mods smoothly
    return a

num1, num2 = 36, 60
print(f"GCD of {num1} and {num2} is: {get_gcd(num1, num2)}")

# Built-in alternative for competitive programming
import math
print(f"Math library GCD alternative: {math.gcd(36, 60)}")
```
