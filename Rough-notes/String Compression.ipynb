```python
def stringCompress(chars):
    size = len(chars)
    idx = 0  # Where we write our compressed answer
    i = 0    # Our reading pointer
    
    while i < size:
        ch = chars[i]
        count = 0
        
        # Count occurrences of the current character
        while i < size and chars[i] == ch:
            count += 1
            i += 1
            
        # Write the character down
        chars[idx] = ch
        idx += 1
        
        # If count > 1, convert count to string and write its digits
        if count > 1:
            for digit in str(count):
                chars[idx] = digit
                idx += 1
                
    # Return the new length of the compressed array
    return idx

# Test data (mimicking C++ vector<char>)
chars = ["a", "a", "b", "b", "c", "c", "c"]
new_length = stringCompress(chars)

# Slice the list up to new_length to see the true result
print("Compressed array:", chars[:new_length]) 
print("Returned length:", new_length)
```
