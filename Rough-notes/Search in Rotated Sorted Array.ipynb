# Search in Rotated Sorted Array

```python
def rotatedArray(arr,target):
    size = len(arr)
    start =1
    end=size-1
    while(start<=end):
        mid= int(start+(end-start)/2)
        if(arr[mid]==target):
            return mid
        if(arr[start]<= arr[mid]):
            if(arr[start]<=target and target<=arr[mid]):
                end=mid-1
            else:
                start = mid+1
        else:
            if(arr[mid]<=target and target <=arr[end]):
                start =mid+1
            else:
                end = mid-1

arr=[4,5,6,7,0,1,2,3]
target = 0
result = rotatedArray(arr,target)
print(f" the element {arr[result]} is present at {result}th index")
```
