```python
def aggressiveCow(arr,cow,size):
    start=min(arr)
    end=max(arr)-start
    ans=-1

    while(start<=end):
        mid=start+(end-start)//2
        if(isValid(arr,mid,size,cow)):
            start=mid+1
            ans=mid
        else:
            end=mid-1
    return ans

def isValid(arr,mid,size,cows):
    cow=1
   
    asc=sorted(arr)
    last_position=asc[0]
    for i in range(size):
        if(asc[i]-last_position>=mid):
            last_position=asc[i]
            cow+=1
        if(cow==cows):

            return True    
    return False
arr=[1,2,8,4,9] 
cow=3
size=len(arr)
result=aggressiveCow(arr,cow,size)
print(result)
```
