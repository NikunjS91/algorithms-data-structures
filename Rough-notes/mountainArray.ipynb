```python
def peakIndex(arr):
    size = len(arr)
    start=1
    end=size-2
    while(start<=end):
        mid= (start+(end-start)//2)
        if(arr[mid-1]<arr[mid] and arr[mid] > arr[mid+1]):
            return mid
        if(arr[mid]<arr[mid+1]):
            start =mid+1
        else:
            end=mid-1
    
    

arr=[0,3,8,9,5,2]
result=peakIndex(arr)
print(f"the peak element is {arr[result]}")
```
