```python
# 'data' is our actual object
data = 10

# 'pointer' points to data
pointer = [data] 

# 'ptr_to_ptr' points to the container holding the pointer
ptr_to_ptr = [pointer]

print(f"Address of Data: {hex(id(data))}")
print(f"Address of Pointer Container: {hex(id(pointer))}")
print(f"Address of Ptr-to-Ptr Container: {hex(id(ptr_to_ptr))}")
```

```python
x = [10, 20]  # x is a reference (pointer) to a list object
print(x)      # Automatically "dereferences" and prints the content
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

node1 = Node(10)
node2 = Node(20)
node1.next = node2

# In C++: node1->next->value
# In Python:
print(node1.next.value) # 20 (Automatic dereferencing of 'next')
```

```python
def try_to_change_value(x):
    print(f"Inside function (start): {x}")
    x = 100  # This creates a NEW local 'x', it doesn't change the original
    print(f"Inside function (end): {x}")

# Main code
my_number = 10
try_to_change_value(my_number)

print(f"Outside function: {my_number}") 
# Output: 10 (The original value did NOT change)
```

```python
def change_my_list(arr):
    print(f"Inside function (start): {arr}")
    arr.append(4) # Modifies the actual object in memory
    arr[0] = "Changed!"
    print(f"Inside function (end): {arr}")

# Main code
my_list = [1, 2, 3]
change_my_list(my_list)

print(f"Outside function: {my_list}") 
# Output: ['Changed!', 2, 3, 4] (The original object WAS modified)
```
