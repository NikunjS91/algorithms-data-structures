```python
def bookAllocation(arr,size,n):
    start=0
    end=sum(arr)
    ans=-1
    while(start<=end):
        mid=start+(end-start)//2
        if(isValid(arr,mid,n,size)):
            end=mid -1
            ans=mid
        else:
            start=mid+1
    return ans

def isValid(arr,maxAllowed,allowedStudent,size):
    student=1
    pages=0
    for i in range(size):
        if(arr[i]>maxAllowed):
            return False
        if(pages+arr[i]<=maxAllowed):
            pages+=arr[i]
        else:
            student+=1
            pages=arr[i]
    if(student>allowedStudent):
        return False
    else:
        return True




arr=[2,1,3,4]
size =len(arr)
n=2
result = bookAllocation(arr,size,n)
print(result)
```
